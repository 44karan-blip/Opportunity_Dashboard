```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix, ConfusionMatrixDisplay

# =====================================================================
# 1. TEMPORAL SIGNUP ANALYSIS (YEAR & MONTH TRENDS)
# =====================================================================
df['Learner SignUp DateTime'] = pd.to_datetime(df['Learner SignUp DateTime'], errors='coerce')
df['Learner SignUp DateTime'] = df['Learner SignUp DateTime'].fillna(pd.Timestamp('2026-01-01'))

df['Signup_Year'] = df['Learner SignUp DateTime'].dt.year
df['Signup_Month'] = df['Learner SignUp DateTime'].dt.strftime('%b')

signup_trends = df.groupby(['Signup_Year', 'Signup_Month']).size().reset_index(name='Signup_Count')
month_order = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
signup_trends['Signup_Month'] = pd.Categorical(signup_trends['Signup_Month'], categories=month_order, ordered=True)
signup_trends = signup_trends.sort_values(['Signup_Year', 'Signup_Month'])

# =====================================================================
# 2. ROBUST FEATURE SELECTION & AUTOMATED TARGET MAPPING
# =====================================================================
feature_cols = ['Opportunity Category', 'Gender', 'Country', 'Current/Intended Major']
target_col = 'Status Description'

# Standardise columns to clean lowercase strings
for col in feature_cols + [target_col]:
    if col in df.columns:
        df[col] = df[col].astype(str).str.strip().str.lower()

# Dynamic Target Mapping to guarantee binary classes (0 and 1)
success_keywords = ['completed', 'success', 'registered', 'active', 'yes', '1', 'approved']
df['Target_Label'] = df[target_col].apply(lambda x: 1 if x in success_keywords else 0)

# SELF-HEALING CHECK: If mapping failed and created only 1 class, fallback to Top Category vs Rest
if df['Target_Label'].nunique() < 2:
    print(f"⚠️ Warning: Keywords didn't create 2 classes. Unique values found: {df[target_col].unique()[:5]}")
    most_common_status = df[target_col].value_counts().index[0]
    print(f"🔄 Automatically setting '{most_common_status}' as 1 and all other statuses as 0.")
    df['Target_Label'] = df[target_col].apply(lambda x: 1 if x == most_common_status else 0)

# Isolate features and target
X = df[feature_cols].copy()
X['Signup_Year_Num'] = df['Learner SignUp DateTime'].dt.year
X['Signup_Month_Num'] = df['Learner SignUp DateTime'].dt.month

y = df['Target_Label']
X_encoded = pd.get_dummies(X, drop_first=True)

# Split data partitions
X_train, X_test, y_train, y_test = train_test_split(X_encoded, y, test_size=0.2, random_state=42, stratify=y)

# =====================================================================
# 3. TRAINING THE PREDICTIVE MODEL
# =====================================================================
model = RandomForestClassifier(n_estimators=100, max_depth=10, random_state=42)
model.fit(X_train, y_train)

y_pred = model.predict(X_test)
print("\n--- Predictive Model Evaluation Summary ---")
print(classification_report(y_test, y_pred))

# Safe probability extraction now that two classes are guaranteed
test_probabilities = model.predict_proba(X_test)[:, 1]
test_results = X_test.copy()
test_results['Success_Probability'] = test_probabilities
test_results['Actual_Outcome'] = y_test.values

# =====================================================================
# 4. GENERATING THE VISUAL GRAPH CANVAS (4 PANELS)
# =====================================================================
fig, axes = plt.subplots(2, 2, figsize=(18, 12))

# Chart 1: Signup Trends Over Time
sns.lineplot(ax=axes[0, 0], data=signup_trends, x='Signup_Month', y='Signup_Count',
             hue='Signup_Year', marker='o', palette='tab10', linewidth=2.5)
axes[0, 0].set_title('Peak Signups Analysis (Year & Month Volumes)', fontsize=12, fontweight='bold')
axes[0, 0].set_xlabel('Month of Year')
axes[0, 0].set_ylabel('Total Registrations')

# Chart 2: Feature Importance Chart
importances = model.feature_importances_
indices = np.argsort(importances)[::-1][:10]
sns.barplot(ax=axes[0, 1], x=importances[indices], y=X_encoded.columns[indices], palette="viridis", hue=X_encoded.columns[indices], legend=False)
axes[0, 1].set_title('Feature Importance Matrix (Top Predictive Drivers)', fontsize=12, fontweight='bold')
axes[0, 1].set_xlabel('Statistical Importance Score')

# Chart 3: Accuracy Confusion Matrix Grid
cm = confusion_matrix(y_test, y_pred)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=['Class 0', 'Class 1'])
disp.plot(ax=axes[1, 0], cmap='Blues', values_format='d')
axes[1, 0].set_title('Model Classification Accuracy Grid', fontsize=12, fontweight='bold')
axes[1, 0].grid(False)

# Chart 4: Risk Cross-Examination Heatmap
encoded_cols = [c for c in X_encoded.columns if 'opportunity category' in c or 'gender' in c or 'country' in c]
feat1 = encoded_cols[0] if len(encoded_cols) > 0 else X_encoded.columns[0]
feat2 = encoded_cols[1] if len(encoded_cols) > 1 else X_encoded.columns[1]

heatmap_data = test_results.pivot_table(
    values='Success_Probability',
    index=feat1,
    columns=feat2,
    observed=False
).fillna(df['Target_Label'].mean())

sns.heatmap(ax=axes[1, 1], data=heatmap_data, annot=True, cmap='RdYlGn', fmt='.2f', cbar_kws={'label': 'Success Probability'})
axes[1, 1].set_title(f'Risk Cohort Analysis Matrix:\n{feat1} vs\n{feat2}', fontsize=11, fontweight='bold')

plt.tight_layout()
plt.show()

# =====================================================================
# 5. EXPORT DASHBOARD DEPLOYMENT ARTIFACTS
# =====================================================================
joblib.dump(model, 'predictive_model.pkl')
joblib.dump(X_encoded.columns.tolist(), 'model_columns.pkl')
test_results.to_excel('opportunity_predictions_matrix.xlsx', index=False)
print("\nModeling Suite Executed successfully! 'opportunity_model.pkl' is ready.")
```
